```python
# Make plots appear inside the notebook
%matplotlib inline

# Import Qiskit and plotting libraries
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# -------------------------------
# Create Quantum Circuit
# -------------------------------
qc = QuantumCircuit(1, 1)   # 1 qubit, 1 classical bit
qc.h(0)                     # Put qubit into superposition
qc.measure(0, 0)            # Measure the qubit

# Draw the circuit
qc.draw('mpl')               # 'mpl' draws using matplotlib

# -------------------------------
# Run Simulation
# -------------------------------
sim = AerSimulator()
result = sim.run(qc, shots=1024).result()
counts = result.get_counts()
print("Measurement counts:", counts)

# -------------------------------
# Plot Histogram
# -------------------------------
plot_histogram(counts)
plt.show()


```

```python
#Pauli-X Gate (Quantum NOT)
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.x(0)              # Flip |0> → |1>
qc.measure(0, 0)

sim = AerSimulator()
result = sim.run(qc, shots=1024).result()
counts = result.get_counts()

print(counts)
```

```python
#Entanglement
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# 1️⃣ Create a 2-qubit, 2-classical-bit circuit
qc = QuantumCircuit(2, 2)

# 2️⃣ Create superposition
qc.h(0)

# 3️⃣ Entangle qubits using CNOT
qc.cx(0, 1)

# 4️⃣ Measure both qubits
qc.measure([0, 1], [0, 1])

# 5️⃣ Show the circuit
print("Quantum Circuit:")
print(qc)

# 6️⃣ Run on simulator
simulator = AerSimulator()
job = simulator.run(qc, shots=1024)
result = job.result()

# 7️⃣ Get measurement counts
counts = result.get_counts()

# 8️⃣ Print results clearly
print("\nMeasurement counts:")
for outcome, count in counts.items():
    print(f"State {outcome} : {count} times")

# 9️⃣ Plot histogram
plot_histogram(counts)
plt.show()
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create circuit: 2 qubits, 1 classical bit
qc = QuantumCircuit(2, 1)

# Step 1: Create entanglement
qc.h(0)        # Superposition
qc.cx(0, 1)    # Entanglement

# Step 2: Measure ONLY qubit 0
qc.measure(0, 0)

# Run on simulator
sim = AerSimulator()
result = sim.run(qc, shots=1024).result()
counts = result.get_counts()

# Show results
print("Measurement counts:", counts)
plot_histogram(counts)
plt.show()
```
